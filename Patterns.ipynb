```python
n=int(input("Enter size:"))
for i in range(n):
    for j in range(n):
        if i==0 or i==n-1 or j==0 or j==n-1:
            print("*", end=' ')
        else:
            print(" ", end=' ')
    print()
```

```python
n=int(input("Enter size:"))
for i in range(n):
    for j in range(n):
        if i==j or i+j==n-1:
            print("*", end=' ')
        else:
            print(" ", end=' ')
    print()
```

```python
n=int(input("Enter size:"))
for i in range(n):
    for j in range(n):
        if i==n//2 or j==n//2:
            print("*", end=' ')
        else:
            print(" ", end=' ')
    print()
```
